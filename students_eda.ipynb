```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv('StudentsPerformance.csv')

print(df.head())
print(df.columns.tolist())

sns.set_style("whitegrid")

# 1. Gender
plt.figure()
sns.countplot(x='gender', data=df)
plt.title('1. Gender Distribution')
plt.show()

# 2. Race
plt.figure()
sns.countplot(x='race/ethnicity', data=df)
plt.title('2. Race Distribution')
plt.show()

# 3. Parental Education
plt.figure()
sns.countplot(y='parental level of education', data=df, order=df['parental level of education'].value_counts().index)
plt.title('3. Parental Education')
plt.show()

# 4. Math Score
plt.figure()
sns.histplot(df['math score'], kde=True)
plt.title('4. Math Score Distribution')
plt.show()

# 5. Reading vs Writing
plt.figure()
sns.scatterplot(x='reading score', y='writing score', hue='gender', data=df)
plt.title('5. Reading vs Writing')
plt.show()

# 6. Lunch vs Math
plt.figure()
sns.boxplot(x='lunch', y='math score', data=df)
plt.title('6. Lunch vs Math Score')
plt.show()

# 7. Test Preparation
plt.figure()
sns.boxplot(x='test preparation course', y='math score', data=df)
plt.title('7. Test Prep vs Math Score')
plt.show()

# 8. Heatmap
plt.figure()
sns.heatmap(df[['math score','reading score','writing score']].corr(), annot=True, cmap='coolwarm')
plt.title('8. Correlation')
plt.show()
```
